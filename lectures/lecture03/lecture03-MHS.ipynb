## Functions

- So far, we have used built-in functions that others wrote: `len()`, `sum()`, ...

Main reasons to use functions:

1.  Code re-use
2.  Robust implementation shielded from other code using clean interfaces
    (inputs and outputs)

### Defining a function

- Use `def` keyword to define own functions
- Function body needs to be *indented*


````python
# Example of keyword def
def func():
    print("Function 'func' was called")
    # Example of indentation
    for i in range(5):
        if i == 1:
            print(f'i is equal to 1')
````

````python
# Call the function
func()
````

### Arguments

- Functions can have an arbitrary number of positional and keyword arguments

### Return values

- `return` statement is used to return a value
- Multiple return values are packed into a `tuple`
- Default return value: `None`
- `return` immediately terminates a function

````python
def increment(x):
    # Return value of x incremented by 1
    return x + 1
````

````python
x = 1
increment(x)
````

### Documenting functions

- Use docstrings (ignored by Python interpreter, but picked up by editors and other tools)

````python
# docstrings (three consecutive quotes) used to document what a function does, its arguments and what it returns
# It displays when you call the function in VS Code

def increment(x):
    """
    Function increments the value of x by 1

    Parameters
    -----------
    x : number
        Number to increment

    Returns
    -------
    number
        x incrementet by one
    """
    return x + 1
````

````python
# Docstring displayed when you call the function
increment()
````

````python
def power(x, xp=2):
    '''
    Function takes x to the power of xp
    '''
    pow = x**xp

    return x, xp, pow

# Default xp
x, xp, pow = power(2)
print(f'{x} to the power of {xp} = {pow}')
# Specify xp as positional argument
x, xp, pow = power(4, 4)
print(f'{x} to the power of {xp} = {pow}')
# Specify xp as keyword argument (can reverse order of arguments, but that requires argument name x as well)
x, xp, pow = power(xp=6, x=4)
print(f'{x} to the power of {xp} = {pow}')
````

#### Arbitrary number of optional arguments

-   `*args`: collects any number of "excess" *positional arguments* and packs
    them into a tuple.
-   `**kwargs`: collects any number of "excess" *keyword arguments* and packs them
    into a dictionary. It needs to be placed at the end of the argument list!

*Example: Function with unnamed positional arguments*

````python
def func(*args):
    '''
    Function that accepts an arbitrary number of positional arguments
    '''
    for i, arg in enumerate(args):
        print(f'Argument {i} is equal to {arg}')


func(1, 2, 3)
````

*Example: Function with unlimited keyword arguments*

````python
# The essential part is the two stars, not the word 'kwargs'. The name can be whatever you like
def func(**kwargs):
    '''
    Function that accepts an arbitrary number of keyword arguments
    '''
    for key, value in kwargs.items():
        print(f'Argument {key} = {value}')


func(argument1 ='value1', argument2=10)
````

*Example: Function with multiple types of arguments*

````python
# Multiple types of arguments
def func(x, opt='default', *args, **kwargs):
    print(f'Required argument x = {x}')
    print(f'Optional argument opt = {opt}')
    # Iterate positional arguments
    for arg in args:
        print(f'positional argument = {arg}')
    # Iterate through keyword arguments
    for key, value in kwargs.items():
        print(f'Keyword argument {key} = {value}')


func(x=10)
func('required', 'optional', 1, 2, 3)  
````

### Working with data from the outer scope

#### Accessing data from the outer scope

- Functions can *read* variables from outer scope (e.g., global module variables)

*Example: print value from outer scope*

````python
x = 1

def func():
    # Attempt to read x from outer (global) scope (it works)
    print(f'The value of x is {x}')


func()
````

````python
x = 2
func()
````

*Example: changed outer scope can affect functions*

````python
x = 1.0

def func():
    x = 2.0  # Retrieves LOCAL variable 'x' (not the same as the GLOBAL 'x'), which equals 2.0
    print(f'The value of x is {x}')

print(x)  # global x
func()  # local x from func()
print(x)  # global x (local and global scope are separate)
````

#### Modifying data in the outer scope

- Cannot write variables from outer scope unless declared with `global` or `nonlocal` (very limited legitimate use cases)
- Usage of outside variables can alter function results, not transparent to caller (avoid this).

*Example: Modifying value from global outer scope*

````python
# NOT RECOMMENDED IN ANY SETTING
x = 1.0

def func():
    global x
    x = 3.0


print(x)  # global x is 1.0
func()  # changes global x to 3.0
print(x)  # global x is 3.0
````

### Pass by value or pass by reference?

**Traditional programming languages (C, Fortran)**

-   *pass-by-value*: a copy of every argument is
    created before it is passed into the function

-   *pass-by-reference*: a reference to a value
    is passed to the function

**Python**

- Passes a copy of a reference to a variable 

*Example: pass scalar value to function*

````python
x = 1.0

def func(x):
    x = 2.0


print(x)
func(x)
print(x)
````

````python
x = 1.0

def func(x):
    x += 1  # ???


print(x)
func(x)  # why is this not the same as when appending to a list???
print(x)
````

*Example: pass list to function*

````python
lst = [1, 2, 3]

def func(lst):
    '''
    Append an element to given list
    '''
    lst.append('hello')


print(lst)
func(lst)  # Function modifies lst from outer scope
print(lst)
````

````python
lst = [1, 2, 3]

def func(lst):
    '''
    Append an element to given list
    '''
    lst = [1, 2, 3, 'hello']


print(lst)
func(lst)   # Function does not modify lst from outer scope, because it assigns something completely different in the function
print(lst)
````
