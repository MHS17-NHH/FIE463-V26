```python
# Libraries
from dataclasses import dataclass
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import root_scalar, minimize_scalar
```

## Part 1

```python
# 1.1. Parameters (tau_a added in part 4, pi_a added in part 6)
@dataclass
class Parameters:
    """
    Model parameters
    """

    beta: float = .99**30           # Discount factor
    gamma: float = 2.1              # Relative risk aversion
    tau: float = .3                 # Payroll tax rate
    phi: float = .5                 # Fraction of wealth bequeathed by old capitalists
    alpha: float = .36              # Capital share in production
    delta: float = 1 - .94**30      # Capital depreciation rate
    z: float = 1.0                  # Total factor productivity
    N_w: float = .8                 # Fraction of workers
    N_k: float = .2                 # Fraction of capitalists
    tau_a: float = .05              # Wealth tax rate (Added in Part 4)
    pi_a: float = .5                # Probability of wealth tax being implemented in the next period (Added in Part 6)
```

```python
# Parameter instance
par = Parameters()
```

```python
# Reusable function to compute prices, labor and capital-labor ratio given capital stock K
def compute_prices(K, par:Parameters):
    """
    Compute prices given capital level K.
    These are derived from the firm's first order conditions.

    Parameters
    ----------
    K : float
        Aggregate capital stock
    par : Parameters
        Model parameters
    
    Returns
    -------
    L : float
        Labor supply
    k : float
        Capital-labor ratio
    w : float
        Wage rate
    r : float
        Interest rate
    """
    # Labor supply (Given and constant in the model as 80% of the population are workers, where 50% are young and active)
    L = .5 * par.N_w

    # Capital-labor ratio (Not directly necessary, but interesting to compare steady-states instances)
    k = K / L
    # Wage
    w = (1 - par.alpha) * par.z * k**par.alpha

    # Interest rate
    r = par.alpha * par.z * k**(par.alpha - 1) - par.delta

    return L, k, w, r
```

```python
# 1.2. Euler error function (With choice of wealth tax added in Part 4)
def euler_err(K_next, K, par: Parameters, wealth_tax=False):
    """
    Compute the capitalists' Euler equation error for a given capital stock today and tomorrow.

    Parameters
    ----------
    K_next : float
        Capital stock in the next period
    K : float
        Capital stock in the current period
    par : Parameters
        Model parameters
    wealth_tax : bool, optional
        Whether wealth tax is implemented (default: False)
    
    Returns
    -------
    euler_error : float
        The Euler equation error; should be zero in equilibrium
    """
    # Wage and rate from Cobb-Douglas production function - This period and next. Function returns L, k, w, r but only r is needed for euler_error!
    *_, r = compute_prices(K, par)
    *_, r_next = compute_prices(K_next, par)

    # Asset market clearing (pre means t-1, t is today)
    a_pre = K / (.5 * par.N_k)
    a_t = K_next / (.5 * par.N_k)

    # Evaluate wealth tax transfer
    tau_a = par.tau_a if wealth_tax else 0

    # Transfers from wealth tax (evaluates to 0 if tau_a = 0) (time subscripts: t for today, t1 for tomorrow)
    T_at = tau_a * (1 + r) * K
    T_at1 = tau_a * (1 + r_next) * K_next

    # Intergenerational transfer based on wealth tax rate and returns on assets previous period
    b_t = par.phi * (1 - tau_a) * (1 + r) * a_pre

    # Euler error consumption values (T's are 0 if no wealth tax, and (1 - tau_a) factor is 1 if no wealth tax, so this works for both cases)
    c_ky = b_t + T_at - a_t                                              
    c_ko = (1 - par.phi) * (1 - tau_a) * (1 + r_next) * a_t + T_at1     

    # Guard against invalid consumption (Might happen at too large K and K_next)
    if c_ky <= 0 or c_ko <= 0:
        # Large positive error: root-finders will avoid this region
        return 1e6

    # Euler error (LHS - RHS)
    LHS = c_ky**(-par.gamma)
    RHS = par.beta * (1 + r_next) * (1 - tau_a) * (1 - par.phi) * c_ko**(-par.gamma)
    euler_error = LHS - RHS
    
    return euler_error
```

```python
# For Part 6.2: Euler error function under uncertainty about wealth tax (Uncertainty enters at expected marginal utility in RHS of Euler equation)
def euler_err_uncertain(K_next, K, par: Parameters, wealth_tax=False):
    """
    Compute the capitalists' Euler equation error under uncertainty about future wealth tax.
    Uncertainty only enters in the RHS of the Euler equation through expected marginal utility.
    Separate function for clarity, but could be combined using conditional statements in euler_err.
    
    Parameters
    ----------
    K_next : float
        Capital stock in the next period
    K : float
        Capital stock in the current period
    par : Parameters
        Model parameters
    wealth_tax : bool, optional
        Whether wealth tax is currently implemented (default: False)
    
    Returns
    -------
    euler_error : float
        The Euler equation error under uncertainty; should be zero in equilibrium
    """
    # Wage and rate from Cobb-Douglas production function - This period and next. The function returns L, k, w, r but only r is needed!
    *_, r = compute_prices(K, par)
    *_, r_next = compute_prices(K_next, par)

    # Asset market clearing (pre means t-1)
    a_pre = K / (.5 * par.N_k)
    a_t = K_next / (.5 * par.N_k)

    # Transfer from wealth tax is certain today, uncertain next period (no-tax case = 0, so it doesn't strictly need to be defined)
    tau_a = par.tau_a if wealth_tax else 0
    T_at = tau_a * (1 + r) * K
    T_at1 = tau_a * (1 + r_next) * K_next

    # Intergenerational transfer is certain
    b_t = par.phi * (1 - tau_a) * (1 + r) * a_pre

    # Consumption values 
    c_ky = b_t + T_at - a_t
    # old age consumption dependent on wealth tax. need to compute both cases for expected marginal utility in RHS of Euler equation                                              
    c_ko_tax = (1 - par.phi) * (1 - tau_a) * (1 + r_next) * a_t + T_at1    
    c_ko_no_tax = (1 - par.phi) * (1 + r_next) * a_t

    # Guard against invalid consumption
    if c_ky <= 0 or c_ko_tax <= 0 or c_ko_no_tax <= 0:
        return 1e6

    # Expected marginal utility of consumption in old age depends on probability of wealth tax being implemented
    exp_term = par.pi_a * (1 - par.tau_a) * c_ko_tax**(-par.gamma) \
          + (1 - par.pi_a) * c_ko_no_tax**(-par.gamma)
   
    # Euler error under uncertainty (LHS - RHS)
    LHS = c_ky**(-par.gamma)
    RHS = par.beta * (1 + r_next) * (1 - par.phi) * exp_term
    euler_error = LHS - RHS
    
    return euler_error
```

```python
# 1.3. Plot euler error for a grid of capital levels K to verify that the function has a root, assuming K_(t+1) = K_t
K_grid = np.linspace(0.001, 0.02, 100)
# Values of K on this grid are very small, but that follows from large delta and 30-year periods.
euler_errors = [euler_err(K, K, par, wealth_tax=False) for K in K_grid]

# Plot Euler error for K_next = K, to verify that the function has a root (i.e. that there is a single K such that Euler error = 0)
plt.plot(K_grid, euler_errors, label='Euler error')
plt.axhline(0, color='k', linestyle='--', label='Zero error')
plt.xlabel('K')
plt.ylabel('Euler error')
plt.title('Euler error for K_next = K')
plt.legend()
_ = plt.grid()
```

The figure shows the Euler equation error as a function of the capital stock K, assuming that capital next period equals capital today $(K_{t+1}=K_t)$. The steady-state capital level corresponds to the point where the Euler error equals zero, meaning that the Euler equation holds exactly. 

The grid for K must be chosen carefully. Because depreciation is very high (one period corresponds to 30 years) and capitalists represent a small share of the population, the equilibrium capital level is very small. If the grid is too large, consumption may become negative or the numerical solution unstable.  

The plot shows that the Euler error crosses zero at around $K≈0.005$, indicating the approximate steady-state level of capital. For lower values of K, the Euler error is negative, implying that capital accumulation is too low relative to the optimal intertemporal allocation. For higher values of K, the Euler error becomes positive, indicating excessive saving.

## Part 2

```python
# 2.1. Define dataclass for steady state values
@dataclass
class SteadyState:
    """
    Storing steady state values.
    Initially empty, but will be filled with values after computing the steady state. 
    """
    par: Parameters = None         # Parameters to compute equilibrium
    K: float = None                # Steady-state aggregate capital
    L: float = None                # Steady-state labor supply
    k: float = None                # Steady-state capital-labor ratio
    w: float = None                # Steady-state wage
    r: float = None                # Steady-state interest rate (This is a 30-year rate)
    r_annual: float = None         # Steady-state ANNUALIZED interest rate
    Y: float = None                # Steady-state output
    c_wy: float = None             # Steady-state consumption of young workers
    c_wo: float = None             # Steady-state consumption of old workers
    a: float = None                # Steady-state asset holdings per young capitalist
    b: float = None                # Steady-state bequests per young capitalist
    c_ky: float = None             # Steady-state consumption of young capitalists
    c_ko: float = None             # Steady-state consumption of old capitalists
    s: float = None                # Steady-state savings rate
    T_a: float = None              # Steady-state wealth tax transfer (per household == aggregate since N == 1)
```

```python
# Reusable function to compute all variables in a given period, clean for steady-state and transition dynamics simulation
def compute_period(K, tau_a, par: Parameters):
    """
    Compute all period variables given capital stock and wealth tax rate.
    Useful for avoiding code repetition in steady-state and transition dynamics.
    Wealth tax rate tau_a will be assigned through other function calls, as compute_period is only used through compute_steady_state and simulate_olg.
    
    Parameters
    ----------
    K : float
        Aggregate capital stock in this period
    tau_a : float
        Wealth tax rate
    par : Parameters
        Model parameters
    
    Returns
    -------
    L : float
        Labor supply
    k : float
        Capital-labor ratio
    w : float
        Wage rate
    r : float
        Interest rate
    T_at : float
        Wealth tax transfer to households
    c_wy : float
        Consumption of young workers
    c_wo : float
        Consumption of old workers
    a : float
        Asset holdings per young capitalist
    b : float
        Bequests per young capitalist
    c_ky : float
        Consumption of young capitalists
    c_ko : float
        Consumption of old capitalists
    s : float
        Savings rate (a / (b + T_at))
    Y : float
        Output
    """
    # Labor, capital-labor ratio, wage, and interest rate from Cobb-Douglas production function
    L, k, w, r = compute_prices(K, par)
    # Wealth tax transfer (per household, which is the same as aggregate since N = 1)
    T_at = tau_a * (1 + r) * K
    # Consumption of young workers
    c_wy = (1 - par.tau) * w + T_at
    # Consumption of old workers (aggregate from Social Security System Balance)
    c_wo = par.tau * w + T_at
    # Asset holdings per young capitalist from market clearing
    a = K / (0.5 * par.N_k)
    # Bequests per young capitalist from intergenerational transfer
    b = par.phi * (1 - tau_a) * (1 + r) * a
    # Consumption of young capitalists
    c_ky = b - a + T_at
    # Consumption of old capitalists
    c_ko = (1 - par.phi) * (1 - tau_a) * (1 + r) * a + T_at
    # Savings rate (of young capitalists)
    s = a / (b + T_at)
    # Output from Cobb-Douglas production function
    Y = par.z * K**par.alpha * L**(1 - par.alpha)
    
    return L, k, w, r, T_at, c_wy, c_wo, a, b, c_ky, c_ko, s, Y
```

```python
# 2.2. Function to compute the steady state calling compute_period. Choice of wealth tax added in Part 4, uncertainty added in Part 6
def compute_steady_state(par: Parameters, wealth_tax=False, uncertainty=False):
    """
    Compute the steady state of the model by finding the capital stock K* where K_next = K.
    
    Parameters
    ----------
    par : Parameters
        Model parameters
    wealth_tax : bool, optional
        Whether wealth tax is implemented (default: False)
    uncertainty : bool, optional
        Whether agents face uncertainty about future wealth tax (default: False)
    
    Returns
    -------
    eq : SteadyState
        Steady-state equilibrium with all variables computed
    """
    # Bracket for root finding, based on the plot of Euler errors
    bracket = (0.001, 0.02)  

    # Define objective function, choices wealth_tax and uncertainty pass through from overall function call, K_next = K for steady state
    def objective(K):
        # We need a different euler error function for uncertainty compared to no uncertainty.
        if uncertainty:
            return euler_err_uncertain(K, K, par)
        else:
            return euler_err(K, K, par, wealth_tax=wealth_tax)
        
    # Find the root of the Euler error function
    result = root_scalar(
        objective,  
        bracket=bracket, 
        method='bisect'
        )
    
    # Check if root finding was successful
    if not result.converged:
        raise ValueError("Root finding did not converge. Try adjusting the bracket or check the Euler error function.")
    
    # Steady-state capital K_star is the root found
    K_star = result.root 

    # Wealth tax or not (Assigned for compute_period function)
    tau_a = par.tau_a if wealth_tax else 0

    # SteadyState instance to store results
    eq = SteadyState(par=par, K=K_star)

    # Compute prices and other variables at the steady state, using compute_period function for cleaner code
    eq.L, eq.k, eq.w, eq.r, eq.T_a, eq.c_wy, eq.c_wo, eq.a, eq.b, eq.c_ky, eq.c_ko, eq.s, eq.Y = compute_period(K_star, tau_a, par)
    eq.r_annual = (1 + eq.r)**(1/30) - 1

    # Goods market clearing check
    lhs = eq.Y + (1 - par.delta) * eq.K
    rhs = .5 * par.N_w * (eq.c_wy + eq.c_wo) + .5 * par.N_k * (eq.c_ky + eq.c_ko + eq.a)

    if abs(lhs - rhs) > 1e-6:
        print("Warning: Goods market clearing condition not satisfied.")
        print(f"LHS: {lhs}, RHS: {rhs}")

    # Returns only the SteadyState instance with all variables at equilibrium
    return eq
```

```python
# Reusable function to print steady state values in a readable format
def print_steady_state(eq: SteadyState, wealth_tax=False, uncertainty=False):
    """
    Print steady-state values in a readable format.
    
    Parameters
    ----------
    eq : SteadyState
        Steady-state equilibrium object
    
    Returns
    -------
    None
        Prints to console 
    """
    print('-------------------------------------------------------------')
    # Heading conditional on wealth tax/uncertainty
    if wealth_tax:
        print("Steady State Values with Wealth Tax:")
    elif uncertainty:
        print("Steady State Values under Uncertainty about Wealth Tax:")
    else:
        print("Steady State Values:")
    print(' Values specified in question:')
    print(f"    Output (Y): {eq.Y:.4f}")
    print(f"    Capital (K): {eq.K:.4f}")
    print(f"    Wage (w): {eq.w:.4f}")
    print(f"    Interest Rate (r): {eq.r:.4%}")
    print(f"    Annualized Interest Rate: {eq.r_annual:.4%}")
    print(f"    Consumption of Young Workers (c_wy): {eq.c_wy:.4f}")
    print(f"    Consumption of Old Workers (c_wo): {eq.c_wo:.4f}")
    print(f"    Consumption of Young Capitalists (c_ky): {eq.c_ky:.4f}")
    print(f"    Consumption of Old Capitalists (c_ko): {eq.c_ko:.4f}")
    print(f"    Asset Holdings per Young Capitalist (a): {eq.a:.4f}")
    print(f"    Savings Rate (s): {eq.s:.4%}")

    print('-------------------------------------------------------------')
    print(' Other values')
    print(f"    Labor (L): {eq.L:.4f} (Constant and Determined by model")
    print(f"    Capital-Labor Ratio (k): {eq.k:.4f}")
    print(f"    Bequests per Young Capitalist (b): {eq.b:.4f}")
    
    # Only print wealth tax transfer if it is implemented (non-zero)
    if eq.T_a == 0:
        print("    No wealth tax implemented.")
    else:
        print(f"    Wealth Tax Transfer per Household (T_a): {eq.T_a:.4f}")

    print('-------------------------------------------------------------')
```

```python
# 2.3. Compute and report steady state values
eq = compute_steady_state(par, wealth_tax=False, uncertainty=False)
print_steady_state(eq)
```

The resulting steady-state capital is approximately $K^*≈0.0049$, which is consistent with the root identified in Part 1. Given this capital level, we compute the implied equilibrium prices, output and consumption levels in the economy. Wage comes to $0.1314$, interest rate (30-year) $516.2944%$, total output $0.0821$, and consumption of workers and capitalists, young and old, $0.0920 (wy), 0.0394 (wo), 0.1025 (cy), 0.1517 (co)$. Capitalists overall consume more, old workers consume more than young, while old capitalists consume the most.

The annualized interest rate is about 6.25%, which seems reasonable given the model calibration. 

## Part 3

```python
# 3.1. Create Simulation dataclass to store time paths of variables
@dataclass
class Simulation:
    par: Parameters                 # Model Parameters
    T: int                          # Number of periods to simulate
    
    K_path: np.ndarray
    r_path: np.ndarray
    w_path: np.ndarray
    Y_path: np.ndarray
    c_wy_path: np.ndarray
    c_wo_path: np.ndarray
    a_path: np.ndarray
    b_path: np.ndarray
    c_ky_path: np.ndarray
    c_ko_path: np.ndarray
    s_path: np.ndarray              # Savings rate s = a/b
```

```python
# 3.2. Function to simulate transition paths, wealth_tax added in Part 4, uncertainty added in Part 6
def simulate_olg(K0, T, par: Parameters, wealth_tax=False, uncertainty=False) -> Simulation:
    """
    Simulate the transition paths of the OLG model starting from initial capital K0 for T periods.
    Uses root-finding to solve for K_{t+1} given K_t using the capitalists' Euler equation.
    
    Parameters
    ----------
    K0 : float
        Initial aggregate capital stock
    T : int
        Number of periods to simulate
    par : Parameters
        Model parameters
    wealth_tax : bool, optional
        Whether wealth tax is implemented (default: False)
    uncertainty : bool, optional
        Whether agents face uncertainty about future wealth tax (default: False)
    
    Returns
    -------
    sim : Simulation
        Simulation object containing time paths of all variables
    """

    # allocate arrays
    K_path   = np.zeros(T)
    r_path   = np.zeros(T)
    w_path   = np.zeros(T)
    Y_path   = np.zeros(T)
    c_wy_path = np.zeros(T)
    c_wo_path = np.zeros(T)
    a_path    = np.zeros(T)
    b_path    = np.zeros(T)
    c_ky_path = np.zeros(T)
    c_ko_path = np.zeros(T)
    s_path    = np.zeros(T)

    # Set initial capital
    K_path[0] = K0
    # Set wealth tax or not
    tau_a = par.tau_a if wealth_tax else 0

    # Simulate every period except last
    for t in range(T - 1):
        # Call compute_period to get all variables for period t
        L, k, w_path[t], r_path[t], T_at, c_wy_path[t], c_wo_path[t], a_path[t], b_path[t], c_ky_path[t], c_ko_path[t], s_path[t], Y_path[t] = compute_period(K_path[t], tau_a, par)

        # Checking goods market clearing inside loop
        lhs = Y_path[t] + (1 - par.delta) * K_path[t]
        rhs = 0.5 * par.N_w * (c_wy_path[t] + c_wo_path[t]) \
        + 0.5 * par.N_k * (c_ky_path[t] + c_ko_path[t] + a_path[t])
        # Prints only if there is a mismatch
        if abs(lhs - rhs) > 1e-8:
            print(f"Goods market mismatch at t={t}: {lhs - rhs}")

        # Define objective function for root finder based on uncertainty or not regarding wealth tax
        def objective(K_next):
            if uncertainty:
                return euler_err_uncertain(K_next, K=K_path[t], par=par)
            else:
                return euler_err(K_next, K=K_path[t], par=par, wealth_tax=wealth_tax)
            
        # Root for K_next
        result = root_scalar(
            objective,
            bracket=(0.001, 0.02),
            method='bisect'
        )
        # Check if root finding was successful
        if not result.converged:
            raise ValueError(f"Root finding did not converge at time {t}.")
        K_path[t+1] = result.root

    # Last period
    L, k, w_path[-1], r_path[-1], T_at, c_wy_path[-1], c_wo_path[-1], a_path[-1], b_path[-1], c_ky_path[-1], c_ko_path[-1], s_path[-1], Y_path[-1] = compute_period(K_path[-1], tau_a, par)

    # Build Simulation object
    sim = Simulation(
        par=par, T=T,
        K_path=K_path, r_path=r_path, w_path=w_path, Y_path=Y_path,
        c_wy_path=c_wy_path, c_wo_path=c_wo_path,
        a_path=a_path, b_path=b_path,
        c_ky_path=c_ky_path, c_ko_path=c_ko_path,
        s_path=s_path
    )
    
    return sim
```

```python
# 3.3. Verify by starting at half the steady-state capital
eq = compute_steady_state(par)
K0 = eq.K / 2  # Start with half of the steady-state capital to see a transition

# SImulate and print the simulated capital path
sim = simulate_olg(K0, T=20, par=par)
print("Simulated capital path:\n", sim.K_path)
```

The simulated path of capital, starting at half the Steady-state, converges nicely back to $K^*$. The simulated model behaves as expected.

```python
# 3.4. Function to plot transition paths
def plot_simulation(eq_old: SteadyState, sim: Simulation, eq_new: SteadyState | None = None, wealth_tax=False, uncertainty=False):
    """
    Plot transition paths for selected variables in a 5x2 subplot grid.
    Variables shown as percent deviations from steady state (except r_t and s_t in percent).
    
    Parameters
    ----------
    eq_old : SteadyState
        Old steady-state equilibrium (used as reference)
    sim : Simulation
        Simulation containing transition paths
    eq_new : SteadyState, optional
        New steady-state equilibrium for comparison (default: None)
    wealth_tax : bool, optional
        Whether wealth tax shock occurred (default: False)
    uncertainty : bool, optional
        Whether uncertainty shock occurred (default: False)
    
    Returns
    -------
    None
        Displays plot
    """
    # Initialize subplots
    fig, ax = plt.subplots(5, 2, figsize=(15, 12))

    # Define variables to plot: (series, title, old SS, new SS if True, plot as % deviation from old SS if True)
    variables = [
        (sim.Y_path,   'Output $Y_t$',   eq_old.Y,   eq_new.Y if eq_new else None, True),
        (sim.K_path,   'Capital $K_t$',  eq_old.K,   eq_new.K if eq_new else None, True),
        (sim.w_path,   'Wage $w_t$',     eq_old.w,   eq_new.w if eq_new else None, True),
        (sim.r_path,   'Interest rate $r_t$', eq_old.r, eq_new.r if eq_new else None, False),
        (sim.b_path,   'Bequests per young capitalist $b_t$', eq_old.b, eq_new.b if eq_new else None, True),
        (sim.s_path,   'Savings rate $s_t$', eq_old.s, eq_new.s if eq_new else None, False),
        (sim.c_wy_path,'Young workers $c^w_{y,t}$', eq_old.c_wy, eq_new.c_wy if eq_new else None, True),
        (sim.c_wo_path,'Old workers $c^w_{o,t}$',   eq_old.c_wo, eq_new.c_wo if eq_new else None, True),
        (sim.c_ky_path,'Young capitalists $c^k_{y,t}$', eq_old.c_ky, eq_new.c_ky if eq_new else None, True),
        (sim.c_ko_path,'Old capitalists $c^k_{o,t}$',   eq_old.c_ko, eq_new.c_ko if eq_new else None, True),
    ]

    # Assign time grid
    T = sim.T
    tgrid = np.arange(T)

    # Loop over variables and plot
    for i, (series, title, ss_old, ss_new, as_pct_dev) in enumerate(variables):
        # Determine subplot location
        row, col = divmod(i, 2)

        # This block plots percent deviations from old/new SS, and old SS as horizontal line at 0
        if as_pct_dev:
            pct_dev = 100 * (series - ss_old) / ss_old
            ax[row, col].plot(tgrid, pct_dev, marker='o', label='Path')
            ax[row, col].axhline(0, color='orange', linestyle='--',
                                 linewidth=1, label='Old SS')
            # Plots new SS in case of wealth tax or uncertainty shock
            if ss_new:
                dev_new = 100 * (ss_new - ss_old) / ss_old
                ax[row, col].axhline(dev_new, color='red', linestyle='--',
                                     linewidth=1, label='New SS')
            ax[row, col].set_ylabel('% deviation from old SS')
        # This block plots percent levels, as well as old SS (and new SS if True) as horizontal lines
        else:
            ax[row, col].plot(tgrid, 100 * series, marker='o', label='Path')
            ax[row, col].axhline(100 * ss_old, color='orange', linestyle='--',
                                 linewidth=1, label='Old SS')
            # Plots new SS in case of wealth tax or uncertainty shock
            if ss_new:
                ax[row, col].axhline(100 * ss_new, color='red', linestyle='--',
                                     linewidth=1, label='New SS')
            ax[row, col].set_ylabel('Percent (%)')
        # Set title, labels, and grid for each subplot
        ax[row, col].set_title(title)
        ax[row, col].set_xlabel('Time')
        ax[row, col].grid(True)

    # Only one legend needed, for path and SS lines by color and style
    ax[0, 0].legend()

    # Overall title dependent on shock type
    if wealth_tax:
        fig.suptitle('Transition paths with wealth tax shock', fontsize=16)
    elif uncertainty:
        fig.suptitle('Transition paths with uncertainty shock, no actual wealth tax', fontsize=16)
    else:
        fig.suptitle('Transition paths without shocks', fontsize=16)

    # Adjust layout and show plot
    plt.tight_layout()
    plt.show()
```

## Part 5

```python
# CRRA Utility function to simplify code and avoid repetition in compute_social_welfare function
def util(c, par:Parameters):
    """
    Compute CRRA utility for given consumption level.
    Handles both the case gamma=1 (logarithmic) and gamma≠1.
    
    Parameters
    ----------
    c : float
        Consumption level
    par : Parameters
        Model parameters (uses gamma for risk aversion)
    
    Returns
    -------
    utility : float
        CRRA utility value
    """
    # CRRA utility function with handling for gamma = 1 case (log utility)
    if par.gamma == 1:
        return np.log(c)
    else:
        return (c**(1 - par.gamma)) / (1 - par.gamma)
```

```python
# 5.1. Compute Social welfare for SS with/without wealth tax
def compute_social_welfare(par: Parameters, wealth_tax=False):
    """
    Compute aggregate social welfare in steady state as population-weighted sum of lifetime utilities.
    
    Parameters
    ----------
    par : Parameters
        Model parameters
    wealth_tax : bool, optional
        Whether wealth tax is implemented (default: False)
    
    Returns
    -------
    SW : float
        Aggregate social welfare
    """
    # Steady-state instance
    eq = compute_steady_state(par, wealth_tax=wealth_tax)
    # CRRA Lifetime Utility of each type using util function
    # Workers
    U_wo = util(eq.c_wo, par)
    U_wy = util(eq.c_wy, par) + par.beta * U_wo
    # Capitalists
    U_ko = util(eq.c_ko, par)
    U_ky = util(eq.c_ky, par) + par.beta * U_ko

    # Population weights
    N_w = eq.par.N_w
    N_k = eq.par.N_k

    # Social welfare as population-weighted sum of utilities
    SW = .5 * N_w * (U_wy + U_wo) + .5 * N_k * (U_ky + U_ko)

    return SW
```

```python
# Compute and print social welfare with and without wealth tax
SW_no_tax = compute_social_welfare(par, wealth_tax=False)
SW_with_tax = compute_social_welfare(par, wealth_tax=True)

print(f'Social welfare without wealth tax: {SW_no_tax:.4f}')
print(f'Social welfare with wealth tax: {SW_with_tax:.4f}')
```

The results show that social welfare is slightly higher when the wealth tax is introduced. This suggests that the average household prefers the economy with the tax. The welfare gain is mainly driven by redistribution from capitalists to workers through the lump-sum transfer.

```python
# 2. Find socially optimal wealth tax rate by maximization as function of tau_a
def max_social_welfare(tau_a, par: Parameters):
    """
    Find the wealth tax rate tau_a that maximizes social welfare.
    Uses bounded scalar minimization on the negative of social welfare.
    
    Parameters
    ----------
    tau_a : float
        Initial wealth tax rate guess
    par : Parameters
        Model parameters
    
    Returns
    -------
    tau_a_opt : float
        Optimal wealth tax rate
    SW_opt : float
        Maximum social welfare at optimal tau_a
    """
    # Define bounds for tau_a (by definition [0, 1])
    bounds = (0, 1)

    # Store original tau_a, to restore later
    tau_a_orig = par.tau_a

    # Objective function to minimize (negative social welfare for maximization)
    def objective(tau_a):
        # Temporarily overwriting tau_a in parameters for social welfare computation
        par.tau_a = float(tau_a)
        SW = compute_social_welfare(par, wealth_tax=True)
        return -SW  

    # Call a minimizer
    res = minimize_scalar(
        objective,
        bounds=bounds,
        method='bounded',
    )

    # Restore original tau_a to parameters
    par.tau_a = tau_a_orig

    # Check if optimization was successful
    if not res.success:
        print('Minimizer exited without success')
        print(res.message)

    # Recover optimal values
    tau_a_opt = res.x
    SW_opt = -res.fun

    return tau_a_opt, SW_opt
```

```python
# Compute and print optimal wealth tax rate and corresponding social welfare
tau_a_opt, SW_opt = max_social_welfare(par.tau_a, par)

print(f'Optimal wealth tax rate (tau_a): {tau_a_opt:.4%}')
print(f'Max social welfare at optimal tau_a: {SW_opt:.4f}')
```

Next, we compute the socially optimal wealth tax rate by maximizing the social welfare function. The optimal tax rate is approximately 4.06%, which is close to the benchmark value used earlier.

```python
# 3. Solve for SS with optimal tau_a, report %-change in Y and K from original SS without wealth tax
eq_old = compute_steady_state(par, wealth_tax=False)

# Separate parameters instance to not overwrite original tau_a with optimal tau_a
par_opt = Parameters()
par_opt.tau_a = tau_a_opt
eq_new = compute_steady_state(par_opt, wealth_tax=True)

# Report %-change in Y and K
pct_change_Y = 100 * (eq_new.Y - eq_old.Y) / eq_old.Y
pct_change_K = 100 * (eq_new.K - eq_old.K) / eq_old.K

print(f'Percent change in output Y from old to new SS: {pct_change_Y:.4f}%')
print(f'Percent change in capital K from old to new SS: {pct_change_K:.4f}%')
```

At this optimal tax rate, output decreases by about 2.12% and capital by about 5.78% compared to the economy without a wealth tax. Despite this decline, welfare increases because redistribution toward households with lower consumption raises total welfare due to diminishing marginal utility of consumption.
